```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

```python

## 1-D Sigmoid
x = np.linspace(-10, 10, 100)
a = 1
y = 1 / (1 + a * np.exp(-x))
plt.plot(x, y)
plt.title('1-D Sigmoid Function')
plt.xlabel('x')
plt.ylabel('sigmoid(x)')
plt.grid()
plt.show()
```

```python
def sigmoid(x, a=1):
    return 1 / (1 + a * np.exp(-x))
```

```python
fig, ax = plt.subplots(2, 3, figsize=(15, 10))
a_values = [0.1, 0.5, 1, 2, 5, 10]
x = np.linspace(-10, 10, 100)
for i, a in enumerate(a_values):
    y = 1 / (1 + np.exp(-1 * a * x))
    ax[i//3, i%3].plot(x, y)
    ax[i//3, i%3].axhline(0.5, color='blue', linestyle='--', linewidth=1)  
    ax[i//3, i%3].set_title(f'1-D Sigmoid Function (a={a})')
    ax[i//3, i%3].set_xlabel('x')
    ax[i//3, i%3].set_ylabel('sigmoid(x)')
    ax[i//3, i%3].grid()
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(2, 3, figsize=(15, 10))   # Create 2x3 subplot grid

a = 1                                             # Sigmoid steepness parameter

f_values = [0.1, 0.5, 1, 2, 5, 10]               # Horizontal shift values

x = np.linspace(-10, 10, 100)                    # Generate 100 x-values from -10 to 10

for i, f in enumerate(f_values):                 # i=index, f=current horizontal shift

    y = 1 / (1 + np.exp(-a * (x - f)))           # Shift sigmoid right by f units

    ax[i//3, i%3].plot(x, y)                     # Plot sigmoid on selected subplot

    ax[i//3, i%3].axhline(0.5, color='blue', linestyle='--', linewidth=1)         # Draw reference line at y=0.5

    ax[i//3, i%3].set_title(f'1-D Sigmoid Function (f={f})')      # Display current shift value

    ax[i//3, i%3].set_xlabel('x')                # Label x-axis

    ax[i//3, i%3].set_ylabel('sigmoid(x)')       # Label y-axis

    ax[i//3, i%3].grid()                         # Show grid

plt.tight_layout()                               # Adjust spacing between plots

plt.show()                                       # Display all subplots
```

```python
fig, ax = plt.subplots(2, 3, figsize=(15, 10))    # Create 2x3 subplot grid

a = 1                                              # Sigmoid steepness

g_values = [0.1, 0.5, 1, 2, 5, 10]                # Vertical shift values

x = np.linspace(-10, 10, 100)                     # 100 x values from -10 to 10

for i, g in enumerate(g_values):                  # i=index, g=current shift value

    y = (1 / (1 + np.exp(-1 * a * x))) + g        # Compute shifted sigmoid

    ax[i//3, i%3].plot(x, y)                      # Plot on correct subplot

    ax[i//3, i%3].axhline(0.5, color='blue', linestyle='--', linewidth=1) # Reference line at y=0.5

    ax[i//3, i%3].set_title(f'1-D Sigmoid Function (g={g})') # Plot title

    ax[i//3, i%3].set_xlabel('x')                 # X-axis label

    ax[i//3, i%3].set_ylabel('sigmoid(x)')        # Y-axis label

    ax[i//3, i%3].grid()                          # Show grid

plt.tight_layout()                                # Adjust spacing

plt.show()                                        # Display all plots
```
