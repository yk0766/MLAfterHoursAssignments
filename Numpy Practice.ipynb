**Calculate mean using loop**

```python
def avg(lst):
    Total = 0
    for nums in lst:
        Total = Total + nums 
    m1 = Total / len(lst)
    return m1
```

**Subtract the mean from every number in the array**

```python
def variance(mean_lst, l1):
    diff = [round((ele - mean_lst),2) for ele in l1]
    return diff
```

```python
a5 = [1, 4, 7, 8]

variance(avg(a5), a5)
```

**Standard Deviation - the calculation using loops**

```python
import math
def std(intgrs):
     sqr_var = [(val - avg(intgrs))**2 for val in intgrs]
     sdev = math.sqrt(sum(sqr_var) / len(intgrs))
     return sdev
```

```python
a9 = [1, 4, 7, 8]

std(a9)
```

**Standard Deviation - the calculation using loops**

```python
Std_arr = np.std(a9)
print(Std_arr)
```

**Fun with sines**

Divide the range  into L intervals

Calculate the sine of these in an ndarray

Plot this the way you’d plot a regular list (using matplotlib.pyplot)

Plot more sine waves as above, but now with different frequencies: odd multiples of the size wave above

```python
# ============================================================
#                  IMPORT LIBRARIES
# ============================================================

import numpy as np
import matplotlib.pyplot as plt


# ============================================================
#               GENERATE SINE WAVE FUNCTION
# ============================================================

# Formula:
# y = A * sin(Bx)
#
# A -> amplitude  (wave height)
# B -> frequency  (number of waves)
#
# Parameters:
# start         -> starting radians value
# end           -> ending radians multiplier of π
# intervals     -> number of intervals between start and end
# frequency     -> controls wave repetition speed
# amplitude     -> controls wave height

def sin_waves(start, end, intervals, frequency, amplitude):

    # Create evenly spaced radians values
    rads = np.linspace(start, end * np.pi, intervals + 1)

    # Generate sine wave values
    sins = amplitude * np.sin(frequency * rads)

    return sins


# ============================================================
#                 PRINT SINE WAVE VALUES
# ============================================================

# sin(x)
print(np.round(sin_waves(0, 2, 4, 1, 1), 2))

# sin(3x)
print(np.round(sin_waves(0, 2, 4, 3, 1), 2))

# 3sin(x)
print(np.round(sin_waves(0, 2, 4, 1, 3), 2))
```

```python
# ============================================================
#                  PLOT SINE WAVES
# ============================================================

# 4 intervals -> creates 5 total points:
# 0, π/2, π, 3π/2, 2π

# Basic sine wave -> sin(x)
plt.plot(sin_waves(0, 2, 4, 1, 1), label="sin(x)")

# Higher frequency wave -> sin(3x)
plt.plot(sin_waves(0, 2, 4, 3, 1), label="sin(3x)")

# Higher amplitude wave -> 3sin(x)
plt.plot(sin_waves(0, 2, 4, 1, 3), label="3sin(x)")
```

```python
# ============================================================
#                  PLOT SINE WAVES
# ============================================================

# Basic sine wave
plt.plot(sin_waves(0, 2, 99, 1, 1), label="sin(x)")

# Higher frequency wave
plt.plot(sin_waves(0, 2, 99, 3, 1), label="sin(3x)")

# Higher amplitude wave
plt.plot(sin_waves(0, 2, 99, 1, 3), label="3sin(x)")


# ============================================================
#                  GRAPH CUSTOMIZATION
# ============================================================

plt.title("Sine Waves with Different Frequencies & Amplitudes")

plt.xlabel("Intervals")

plt.ylabel("Wave Height")

plt.legend()

plt.show()
```
