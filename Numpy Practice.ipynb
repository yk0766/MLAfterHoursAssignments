```python
n = 10


lst = [random.randint(10, 20) for i in range(n)]
arr = np.random.randint(10, 20, size = n)

print(lst)
print(arr)
```

**Calculate mean using loop**

```python
def avg(lst):
    Total = 0
    for nums in lst:
        Total = Total + nums 
    m1 = Total / len(lst)
    return m1
```

**Subtract the mean from every number in the array**

```python
def variance(mean_lst, l1):
    diff = [round((ele - mean_lst),2) for ele in l1]
    return diff
```

```python
a5 = [1, 4, 7, 8]

variance(avg(a5), a5)
```

**Standard Deviation - the calculation using loops**

```python
import math
def std(intgrs):
     sqr_var = [(val - avg(intgrs))**2 for val in intgrs]
     sdev = math.sqrt(sum(sqr_var) / len(intgrs))
     return sdev
```

```python
a9 = [1, 4, 7, 8]

std(a9)
```

**Standard Deviation - the calculation using loops**

```python
Std_arr = np.std(a9)
print(Std_arr)
```
