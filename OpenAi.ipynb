```python
# Import load_dotenv so Python can read variables from the .env file
from dotenv import load_dotenv

# Import os so we can access environment variables
import os

# Load the variables stored inside the .env file
load_dotenv()

# Get the value of OPENAI_API_KEY from the environment
api_key = os.getenv("OPENAI_API_KEY")

# Check whether the API key was successfully loaded
# This prints True or False without exposing the actual key
print("API key loaded:", bool(api_key))
```

```python
from openai import OpenAI   # Get the OpenAI tool

client = OpenAI()           # Create the client we will use to talk to OpenAI

# Ask a question to the OpenAI model
response = client.responses.create(
    model="gpt-5.6",
    input="What is machine learning? Explain it in simple terms."
)

# Print the model's answer
print(response.output_text)
```

```python
# Create an empty list to store all questions and answers
chat_history = []

while True:

    # Ask the user to enter a question
    question = input("You: ")

    # Stop the loop if the user types "exit"
    if question.lower() == "exit":
        break

    # Send the question to OpenAI
    response = client.responses.create(
        model="gpt-5.6",
        input=question
    )

    # Get the answer from the response
    answer = response.output_text

    # Store both the question and answer
    chat_history.append({
        "question": question,
        "answer": answer
    })

    # Display the answer
    print("ChatGPT:", answer)
```
